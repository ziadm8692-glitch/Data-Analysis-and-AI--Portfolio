# ***Feature Engineering***

```python
def for_periods_of_time (hr):
  if hr>=0 and hr<6:
    return 'Night'
  elif hr>=6 and hr<12:
    return 'Morning'
  elif hr>=12 and hr<18:
    return 'Afternoon'
  else:
    return 'Evening'

df['period_of_day'] = df['hr'].apply(for_periods_of_time)
```

```python
# Key Insights
"""
Domain-Driven Categorization: Grouping hours into logical daylight segments reduces hourly granularity while maintaining high-level temporal patterns (e.g., commute vs. leisure hours).
Feature Utility: Converts a continuous/discrete integer hour (0-23) into an impactful categorical feature (period_of_day),
helping non-linear and linear models capture non-monotonic daily demand fluctuations efficiently
"""
```

```python
df['day_of_week']=df['dteday'].dt.day_of_week
df.head()
```

```python
# Key Insight
"""
Temporal Granularity: Adds a critical cyclical feature (0 to 6) allowing models to capture specific weekly patterns and weekend lifestyle shifts.
 Feature Integration: Successfully appends the new column alongside previous transformations (Season, Year, period_of_day),
 ensuring a clean and enriched dataset across all 23 columns.
"""
```

```python
df['my_calculated_weekend'] = df['day_of_week'].isin([5, 6]).astype(int)
print(df[['dteday', 'day_of_week', 'Workingday', 'my_calculated_weekend']].head(10))
```

```python
# Key Insights
"""
Automated Flagging: Effectively isolates weekend records from weekdays to capture distinct cyclical rental patterns.
 Data Validation: Output verification confirms that January 1, 2011 (2011-01-01),
  mapped correctly as day 5 (Saturday),
registering as a non-working day with a weekend value of 1.
"""
```

# ***Data Preprocessing***

```python
df = pd.get_dummies(df, columns=['season', 'holiday', 'workingday', 'period_of_day'], drop_first=True)
```

```python
df.drop(columns=['Season', 'Holiday', 'Workingday','instant'], inplace=True, errors='ignore')
```

```python
df.columns
```

```python
# Key Insight
"""
Multicollinearity Prevention: Using drop_first=True removes the dummy variable trap, reducing correlation between encoded features.
Feature Standardization: Converts all categorical columns into binary numeric representations (0 or 1) ready for algorithm training.
"""
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scale_cols = ['atemp', 'temp', 'hum', 'windspeed', 'casual', 'registered', 'cnt']

encoding = StandardScaler()
df[scale_cols] = encoding.fit_transform(df[scale_cols])
```

```python
# Key Inisght
"""
Zero Mean & Unit Variance: Transforms features to have a mean ($\mu$) of 0 and a standard deviation ($\sigma$) of 1,
preventing high-magnitude features from dominating model distance calculations.
Critical Pipeline Warning: Note that scaling cnt (target variable) and leakage variables (casual, registered) should be handled carefully before model splitting to avoid data leakage
"""
```

```python
df.head()
```

```python
df['Date_Numeric'] = pd.to_datetime(df['dteday']).map(pd.Timestamp.toordinal)
df.drop(columns=['dteday'], inplace=True)
```
